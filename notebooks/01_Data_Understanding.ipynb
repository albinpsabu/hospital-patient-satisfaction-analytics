 Project Objectives

2. Business Problem

Hospitals collect large amounts of patient-flow and patient-experience
data. However, raw data alone does not clearly explain what factors are
associated with patient satisfaction.

This project aims to analyze patient experience and operational data
to identify important patterns, understand potential factors associated
with satisfaction, and develop a predictive analytics solution that can
support hospital management.
Then add one more Markdown cell: Business

3. Business Questions

The analysis will answer the following questions:

1. What is the overall distribution of patient satisfaction?

2. Which departments or patient-flow categories show differences
   in satisfaction?

3. Is waiting time associated with patient satisfaction?

4. Which operational or service-related variables are associated
   with satisfaction?

5. Are there significant differences in satisfaction between
   different patient groups?

6. Are there outliers or unusual patterns in patient-flow data?

7. Which factors are most useful for predicting patient satisfaction?

8. Can a machine learning model identify patients who may have
   lower satisfaction?

9. What actionable insights can hospital management obtain from
   the analysis?

4. Data Dictionary

| Column | Data Type | Description | Role |
|---|---|---|---|
| Patient Id | Object | Unique identifier for each patient record | Identifier |
| Patient Admission Date | Object | Date on which the patient was admitted | Date |
| Patient Admission Time | Object | Time of patient admission | Time |
| Merged | Object | Combined admission date/time field | Date-Time |
| Patient Gender | Object | Gender of the patient | Categorical |
| Patient Age | Integer | Age of the patient | Numerical |
| Patient Race | Object | Race/ethnicity category recorded in the dataset | Categorical |
| Department Referral | Object | Department associated with the patient encounter | Categorical |
| Patient Admission Flag | Object | Indicates the admission/visit status recorded in the dataset | Categorical |
| Patient Satisfaction Score | Float | Patient satisfaction score recorded after the encounter | Target / Numerical |
| Patient Waittime | Integer | Patient waiting time | Numerical |

### Initial Analytical Roles

- **Target variable:** Patient Satisfaction Score

- **Categorical variables:** Patient Gender, Patient Race, Department Referral, Patient Admission Flag
- **Date/Time variables:** Patient Admission Date, Patient Admission Time, Merged
- **Identifier:** Patient Id
