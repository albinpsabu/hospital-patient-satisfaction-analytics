# Hospital Patient Satisfaction Analytics & Prediction

## 2. Feature Engineering and Preprocessing

This notebook prepares the cleaned dataset for machine learning.

### Objectives

- Load the processed dataset
- Define the prediction target
- Analyze target availability and distribution
- Select relevant features
- Handle missing feature values
- Encode categorical variables
- Apply feature scaling where required
- Prepare the final dataset for machine learning

```python
import pandas as pd
import numpy as np

processed_path = "/content/drive/MyDrive/datasets2026/healthcare_patient_flow_cleaned.csv"

df = pd.read_csv(processed_path)

print("Processed dataset loaded successfully")
print("Shape:", df.shape)
```

```python
# after loading the processed path we want to define the target
target = "Patient Satisfaction Score"

print("Target variable:", target)
```

```python
#then we want to define the model_df
model_df = df.dropna(subset=[target]).copy()

print("Modeling dataset shape:", model_df.shape)
```

```python
#Then print and verify the columns
print("Total records:", len(df))
print("Missing target:", df[target].isna().sum())
print("Available target:", df[target].notna().sum())
print("Modeling records:", len(model_df))
```

```python
# then we want to define the numerical column and the categorical column
numerical_features = [
    "Patient Age",
    "Patient Waittime"
]

categorical_features = [
    "Patient Gender",
    "Patient Race",
    "Department Referral",
    "Patient Admission Flag"
]

target = "Patient Satisfaction Score"

print("Numerical features:", numerical_features)
print("Categorical features:", categorical_features)
print("Target:", target)
```

```python
# then we want to create the x and y so that the :
#the x contains the model of numerical_features and the categorical features
x=model_df[numerical_features +categorical_features]
y=model_df[target]
print("x shape:",x.shape)
print("y shape:",y.shape)
```

```python
# ok now we want to split the  train data  and test data
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.20,random_state=42
)
print("X_train:",x_train.shape)
print("X_test:",x_test.shape)
print("y_train:",y_train.shape)
print("y_test:",y_test.shape)

```

```python
#ok next we want to do the define of simpleImputer
from sklearn.impute import SimpleImputer# it is used for filling
#the missing values in the dataset by NAN
from sklearn.preprocessing import OneHotEncoder#it is used for converting the
#categorical text data into numerical data
from sklearn.pipeline import Pipeline#it is used to execute the multiple preprocessing in
# a fixed order
from sklearn.compose import ColumnTransformer # it is used to choose which columns will do the preprocessing
```

```python
#then we want to define the numerical Transformer
numerical_transformer=Pipeline(
    steps=[
        ("imputer",SimpleImputer(strategy="median"))
    ]
)
```

```python
#then we want to define the categorical transformer
categorical_transformer=Pipeline(
    steps=[
        ("imputer",SimpleImputer(strategy="most_frequent")),
        ("encoder",OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

```python
preprocessor=ColumnTransformer(
    transformers=[
        (
            "num",
            numerical_transformer,
            numerical_features
        ),
        (
           "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)
print("Preprocessor created successfully")
```

```python
# ok now we have to do the fit_transform
#fitransform means it is the combination of two operations
#fit means it will decide which data has to be changed from the rule to the missing values
#transform means it will apply the changes to the data and transform it
x_train_processed=preprocessor.fit_transform(x_train)
x_test_processed=preprocessor.transform(x_test)

print("Training data processed successfully")
print("Testing data processed data successfully")


```

```python
#ok now we have to check the shape
print("Original x_train shape:",x_train.shape)
print("proccessed x_train shape",x_train_processed.shape)

print("Original x_test shape:",x_test.shape)
print("processed x_test shape:",x_test_processed.shape)
```

```python
#ok now want to know ,after encoding  how many new columns are created newly
feature_names=preprocessor.get_feature_names_out()
print("Feature names:")
print(feature_names)
print(len(feature_names))
```

```python
# ok now we have to store the processed data to a dataframe
x_train_processed_df=pd.DataFrame(x_train_processed,
                                  columns=feature_names,
                                  index=x_train.index)
x_test_processed_df=pd.DataFrame(x_test_processed,
                                 columns=feature_names,
                                 index=x_test.index)
print("Processed training data:")
print(x_train_processed_df.head())
```

```python
#Ok now we have to find the shape of the proceesed train data and test data
print(x_test_processed_df.shape)
print(x_train_processed_df.shape)

# now we want to check if any missing values are there after preprocessing
print(x_train_processed_df.isna().sum().sum())
print(x_train_processed_df.head())
```

```python
#Ok now we want to check the datatypes of the dataset
print(x_train_processed_df.dtypes.value_counts())
```

```python
# ok atlast we want to check if the target and the processed features are ready
print(x_train_processed_df.shape)
print(x_test_processed_df.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
#ok now we want to save the current processed and datasets
import joblib
preprocessor_path = "/content/drive/MyDrive/hospital_preprocessor.pkl"

joblib.dump(
    preprocessor,
    preprocessor_path
)

print("Preprocessor saved successfully.")
```

```python
train_data_path = "/content/drive/MyDrive/hospital_X_train_processed.csv"
test_data_path = "/content/drive/MyDrive/hospital_X_test_processed.csv"

x_train_processed_df.to_csv(
    train_data_path,
    index=False
)

x_test_processed_df.to_csv(
    test_data_path,
    index=False
)

print("Processed training and testing data saved.")
```

```python
y_train.to_csv(
    "/content/drive/MyDrive/hospital_y_train.csv",
    index=False
)

y_test.to_csv(
    "/content/drive/MyDrive/hospital_y_test.csv",
    index=False
)

print("Target data saved.")
```
